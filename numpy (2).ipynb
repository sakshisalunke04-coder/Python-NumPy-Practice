```python
import numpy as np
salary=np.array([25000,32000,33000,40000,500000,45000,56000,60000,65000])
print("mean" ,np.mean(salary))
print("median",np.median(salary))
print("average",np.average(salary))
print("standard deviation",np.std(salary))
print("variance",np.var(salary))
print("minimum",np.min(salary))
print("maximum",np.max(salary))
print("total",np.sum(salary))
avg=np.mean(salary)
print("above average",salary[salary>avg])
print("below average",salary[salary<avg])
print("salary between 30000 & 50000",salary[(salary>=30000)&(salary<=50000)])
print("heighest salary",np.max(salary))
print("lowest")

```
